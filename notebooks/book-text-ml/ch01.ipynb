```python
# 1. 모든 컬럼의 결측치(비어있는 값) 개수 확인
print("전체 컬럼 결측치 개수:")
print(df_books.isna().sum())

# 2. 결측치가 1개 이상 있는 컬럼만 필터링해서 보기
null_counts = df_books.isna().sum()
print("\n결측치가 있는 컬럼만 보기:")
print(null_counts[null_counts > 0])
```

### 실습 6. 판매가를 숫자로 바꾸기

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다. 

16,200

17,820

7,200
 
사람이 보기에는 숫자처럼 보이지만 pandas에서는 문자열로 읽힐 수 있습니다.

문자열 상태에서는 평균, 합계 등의 숫자 계산을 제대로 할 수 없습니다.

따라서 쉼표를 제거하고 숫자 자료형으로 변환합니다.

먼저 자료형 확인

df_books["판매가"].dtype

```python
# 1. '판매가' 컬럼에서 쉼표(,) 제거하기
df_books['판매가'] = df_books['판매가'].str.replace(',', '')

# 2. 문자열(str) 형태를 정수형(int)으로 변환하기
df_books['판매가'] = df_books['판매가'].astype(int)

# 3. 변환 후 '판매가' 컬럼의 자료형(dtype) 확인하기
print("판매가 컬럼 자료형:", df_books['판매가'].dtype)

# 4. 상위 5행 출력하여 값 확인하기
df_books.head()
```

### 판매가 전처리 실행 결과 요약

* **전처리 내용:** `.str.replace(',', '')`를 이용해 쉼표를 제거한 후, `.astype(int)`로 정수형 변환 완료
* **자료형 확인:** `판매가` 컬럼의 자료형이 문자열(`object`)에서 정수형(`int64`)으로 성공적으로 변경됨
* **데이터 확인:** 상위 5행 출력 결과, `'13,500'` $\rightarrow$ `13500`과 같이 쉼표가 제거된 숫자 데이터로 정상 변환된 것 확인

### 실습 7. 발행일을 날짜형으로 바꾸기

 

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

 

20260701

20260831

20020120

 

이 값은 연도-월-일 정보를 가지고 있지만 바로 날짜 계산에 사용하기 어려울 수 있습니다.

 

날짜 자료형으로 변환하면 연도, 월, 날짜 등을 쉽게 분석할 수 있습니다.


```python
# 1. '발행일' 컬럼을 문자열(str)로 변환 후, YYYYMMDD 포맷에 맞게 datetime 날짜형으로 변환
df_books['발행일'] = pd.to_datetime(df_books['발행일'].astype(str), format='%Y%m%d')

# 2. 변환 후 자료형(dtype) 확인
print("발행일 컬럼 자료형:", df_books['발행일'].dtype)

# 3. 앞의 5개 값 확인 (상위 5행)
df_books[['상품명', '발행일']].head()
```

### 발행일 전처리 실행 결과 요약

* **전처리 내용:** `.astype(str)`로 문자열 변환 후 `pd.to_datetime(..., format='%Y%m%d')`를 적용해 날짜 형식으로 변환 완료
* **자료형 확인:** `발행일` 컬럼의 자료형이 정수형(`int64`)에서 날짜형(`datetime64[us]`)으로 성공적으로 변경됨
* **데이터 확인:** 상위 5행 출력 결과, `20140519` $\rightarrow$ `2014-05-19`와 같이 `YYYY-MM-DD` 형태로 정확히 변환된 것을 확인

### 실습 8. 문자열 앞뒤 공백 정리하기

 

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

 

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

 

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.

 

"소설"

"소설 "

 

이번에 정리할 컬럼

 

상품명

저자

출판사

분야

```python
# 1. 공백을 제거할 문자열 컬럼 목록 지정
str_cols = ['상품명', '저자', '출판사', '분야']

# 2. 반복문을 사용해 각 컬럼의 앞뒤 공백 제거 (str.strip() 적용)
for col in str_cols:
    df_books[col] = df_books[col].astype(str).str.strip()

# 3. 전처리 후 해당 컬럼들의 상위 5행 데이터 확인
df_books[str_cols].head()
```

### 문자열 공백 제거 실행 결과 요약

* **대상 컬럼:** `상품명`, `저자`, `출판사`, `분야` (총 4개)
* **전처리 내용:** 반복문(`for`)을 활용해 지정한 컬럼들의 텍스트 데이터에 `.astype(str).str.strip()`을 일괄 적용
* **결과 확인:** 문자열 양끝의 불필요한 공백이 성공적으로 제거되었으며, `df_books[str_cols].head()`를 통해 상위 5행 데이터가 정상 정제되었음을 확인

### 실습 9. 중복 데이터 확인하기

 

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

 

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

 

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.

 

하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

 

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

 

우선 상품을 구분하는 판매상품ID를 확인합니다.
 

<중요한 생각>

 

상품명이 같다

≠ 반드시 잘못된 중복 데이터다

 

중복 데이터는 먼저 의미를 확인한 뒤 처리합니다.

 

이번 단계에서는 중복 여부만 확인하고 원본 행을 임의로 삭제하지 않습니다.

 

```python
# 1. 전체 행이 완전히 똑같은 중복 데이터 개수 확인
duplicated_all = df_books.duplicated().sum()
print("전체 행 중복 개수:", duplicated_all)

# 2. '판매상품ID'가 중복된 행 개수 확인
duplicated_id = df_books.duplicated(subset=['판매상품ID']).sum()
print("판매상품ID 중복 개수:", duplicated_id)

# 3. '상품명'이 중복된 행 개수 확인
duplicated_name = df_books.duplicated(subset=['상품명']).sum()
print("상품명 중복 개수:", duplicated_name)
```

### 중복 데이터 확인 실행 결과 요약

* **전체 행 중복 개수:** 0건
* **판매상품ID 중복 개수:** 0건
* **상품명 중복 개수:** 0건
* **결론:** `duplicated()` 함수 검사 결과, 전체 행뿐만 아니라 고유 키 역할을 하는 `판매상품ID` 및 `상품명` 기준에서도 중복 데이터가 0건으로 확인되어 데이터의 유일성이 잘 유지되고 있습니다.
